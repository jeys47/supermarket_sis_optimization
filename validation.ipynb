```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration globale du style
sns.set_theme(style="whitegrid")

# ==============================================================================
# 1. CHARGEMENT ET SÉLECTION DES COLONNES DEMANDÉES
# ==============================================================================
try:
    df = pd.read_csv("dataset_cleaned.csv")
except NameError:
    df = df.copy()

# Nettoyage des noms de colonnes (suppression des espaces)
df.columns = [c.strip() for c in df.columns]

# Liste exacte des colonnes souhaitées
colonnes_cible = [
    'produit', 'categorie', 'sous_categorie', 'marque', 
    'rayon_id', 'rayon', 'position_rayon', 'niveau_rayon', 
    'quantite', 'prix_unitaire', 'prix_total'
]

# Gestion des colonnes manquantes (si certaines ne sont pas explicites dans le CSV)
for col in colonnes_cible:
    if col not in df.columns:
        if col == 'produit':
            df['produit'] = "Produit " + df['niveau_rayon'].astype(str) + " (" + df['prix_unitaire'].astype(str) + "$)"
        elif col == 'categorie':
            df['categorie'] = df['rayon']
        elif col == 'sous_categorie':
            df['sous_categorie'] = df['position_rayon']
        elif col == 'marque':
            df['marque'] = "Marque Générique"

# ==============================================================================
# 2. RANKING DÉTAILLÉ DE TOUS LES PRODUITS (TOP 20)
# ==============================================================================
ranking_produits = df.groupby([
    'produit', 'rayon', 'rayon_id', 'categorie', 'sous_categorie', 
    'marque', 'position_rayon', 'niveau_rayon', 'prix_unitaire'
]).agg(
    quantite_totale=('quantite', 'sum'),
    prix_total_cumule=('prix_total', 'sum')
).reset_index()

# Tri par quantité vendue (ordre décroissant)
ranking_produits = ranking_produits.sort_values(by='quantite_totale', ascending=False).reset_index(drop=True)
ranking_produits.index = ranking_produits.index + 1
ranking_produits.index.name = 'Rang'

print("=" * 100)
print(" TOP 20 DES MEILLEURS PRODUITS (AVEC DETAILS COMPLET)")
print("=" * 100)
display(ranking_produits.head(20)[
    ['produit', 'rayon', 'categorie', 'sous_categorie', 'marque', 
     'position_rayon', 'niveau_rayon', 'quantite_totale', 'prix_unitaire', 'prix_total_cumule']
])

# ==============================================================================
# 3. RANKING DES MEILLEURS RAYONS
# ==============================================================================
ranking_rayons = df.groupby(['rayon_id', 'rayon']).agg(
    quantite_totale=('quantite', 'sum'),
    chiffre_affaires_total=('prix_total', 'sum')
).reset_index().sort_values(by='quantite_totale', ascending=False).reset_index(drop=True)


```
